# Source flow: the "Try it!" worked from the beginning

Companion to [Lesson 1 - Source & Sink](../01_Lesson01_sourceSink.ipynb).

**Question:** if fluid flows straight outward from a point, why does its speed decrease as the inverse of the distance from that point?

We will first explain the coordinates and velocity notation. Then we will use two physical conditions: fluid elements do not spin (**irrotational flow**), and fluid is conserved (**continuity**). Finally, we will define the source strength and obtain the formula in the lesson.

## 1. Where are we measuring the velocity?

Put the source at the origin, the point $(x,y)=(0,0)$. Pick another point $P$ in the plane.

There are two ways to describe the position of $P$:

- **Cartesian coordinates:** $x$ is its horizontal position and $y$ is its vertical position.
- **Polar coordinates:** $r$ is its distance from the origin, and $\theta$ (the Greek letter *theta*) is the angle from the positive horizontal axis, measured counterclockwise in radians.

They are related by

$$x=r\cos\theta,\qquad y=r\sin\theta,\qquad r=\sqrt{x^2+y^2}.$$

For example, $(x,y)=(1,0)$ has $r=1$ and $\theta=0$. The point $(0,1)$ has $r=1$ and $\theta=\pi/2$; $\pi$ is the usual circle constant, approximately $3.14159$.

The lesson calls these cylindrical coordinates. Because the flow is two-dimensional, we only need the radius and angle; there is no variation in the third direction.

We work at $r>0$: the ideal point source itself will be a singular point where the formula is undefined.

## 2. What are the bold e symbols?

A **vector** has a magnitude and a direction. A **unit vector** is a vector of length one: it specifies a direction without setting the speed.

At the point $P$, we define two unit vectors:

- $\mathbf{e}_r$: an arrow of length one pointing **straight away from the origin**. The subscript $r$ labels the radial direction.
- $\mathbf{e}_\theta$: an arrow of length one pointing **along a circle around the origin**, toward increasing $\theta$ (counterclockwise). It is perpendicular to $\mathbf{e}_r$.

Here **$\mathbf{e}$ is a conventional name for a direction vector**. It is not Euler's number $e\approx2.718$, an exponent, or a velocity. The bold type indicates a vector; the subscript tells us which direction it represents.

Their horizontal and vertical components are

$$\mathbf{e}_r=(\cos\theta,\sin\theta),\qquad
\mathbf{e}_\theta=(-\sin\theta,\cos\theta).$$

For a concrete picture:

| Position of $P$ | $\mathbf{e}_r$ points | $\mathbf{e}_\theta$ points |
|---|---|---|
| Right of the source, $\theta=0$ | Right: $(1,0)$ | Up: $(0,1)$ |
| Above the source, $\theta=\pi/2$ | Up: $(0,1)$ | Left: $(-1,0)$ |

These directions change as we move around the source. They are directions defined **at the point where we measure the flow**.

## 3. What do the velocity symbols mean?

Let $\mathbf{u}$ be the fluid velocity vector at $P$. Its units are metres per second. We describe it using two signed components:

- $u_r$: the velocity in the radial direction. Positive means outward; negative means inward.
- $u_\theta$: the velocity in the tangential direction. Positive means counterclockwise; negative means clockwise.

The full velocity vector is

$$\mathbf{u}=u_r\mathbf{e}_r+u_\theta\mathbf{e}_\theta.$$

Read this as: **velocity = radial speed times radial direction + tangential speed times tangential direction**. Multiplying a unit vector by a signed speed gives the velocity contribution in that direction.

For example, at a point to the right of the source, $u_r=2$ and $u_\theta=0$ mean the fluid moves right at $2\ \mathrm{m/s}$. At a point above the source, the same components mean it moves up at $2\ \mathrm{m/s}$.

For the pure source described in the lesson, the fluid moves along straight lines away from the origin. It has no motion around the origin, so

$$u_\theta=0.$$

We therefore only need to find $u_r$. Initially, write it as $u_r(r,\theta)$, meaning it might depend on distance and angle. We assume **steady** flow, so it does not depend on time.

## 4. The derivative notation we will use

A derivative measures how one quantity changes when another changes.

The symbol $\partial$ is used for a **partial derivative**, when a function can depend on more than one variable:

- $\partial u_r/\partial\theta$ measures how radial velocity changes as we change the angle while holding the radius fixed: we move around a circle.
- $\partial u_r/\partial r$ measures how radial velocity changes as we change the radius while holding the angle fixed: we move along a ray.

Thus $\partial u_r/\partial\theta=0$ means that radial velocity is the same at every angle on a given circle.

If a function depends only on $r$, we can use the ordinary derivative symbol $d/dr$ instead of $\partial/\partial r$.

## 5a. Where does the polar vorticity formula come from?

We need to derive the formula, not assume it. Start with a way to measure local rotation: **circulation around a tiny closed boundary, divided by the area inside it**.

### Circulation means adding the velocity along a loop

Walk counterclockwise around a closed boundary. On each short piece, multiply the fluid velocity **along your walking direction** by the length of that piece, then add all the contributions. Motion along your walk counts positive; motion against it counts negative. This sum is called circulation, written $\Gamma$ (capital Greek *gamma*). Its units are $\mathrm{m^2/s}$.

The perpendicular component of vorticity is defined locally by

$$\omega_z=\lim_{\text{patch shrinks to a point}}\frac{\Gamma}{\text{patch area}}.$$

The limit means taking progressively smaller patches. Counterclockwise circulation is positive, corresponding to the positive $z$ direction out of the page. Vorticity has units $1/\mathrm{s}$; it is twice the local angular rotation rate of a fluid element. For example, a rigidly rotating disk with angular speed $\Omega$ has tangential speed $\Omega a$ on a circle of radius $a$: its circulation divided by area is $(2\pi a)(\Omega a)/(\pi a^2)=2\Omega$.

### Choose a small patch that follows polar coordinates

Let $\Delta r$ be a small positive radial width and $\Delta\theta$ a small positive angular width. The symbol $\Delta$ means a finite increment. Our patch lies between radii $r$ and $r+\Delta r$, and angles $\theta$ and $\theta+\Delta\theta$.

Its four corners, listed counterclockwise, are

$$A=(r,\theta),\quad B=(r+\Delta r,\theta),\quad
C=(r+\Delta r,\theta+\Delta\theta),\quad D=(r,\theta+\Delta\theta).$$

These ordered pairs are **polar** coordinates. Walk $A\to B\to C\to D\to A$: outward, around the outer arc, inward, then back along the inner arc. The two radial sides have length $\Delta r$. The inner arc has length $r\Delta\theta$ and the outer arc has length $(r+\Delta r)\Delta\theta$.

The exact area of this annular sector is

$$\text{area}=\frac{(r+\Delta r)^2-r^2}{2}\Delta\theta
=\left(r\Delta r+\frac{(\Delta r)^2}{2}\right)\Delta\theta.$$

For a shrinking patch, its leading area is $r\Delta r\Delta\theta$.

### Add the four signed contributions

To leading order for a small patch, each contribution is the velocity along that side times its length:

| Side | Walking direction | Contribution to circulation |
|---|---|---|
| $A\to B$ | Outward, at angle $\theta$ | $+u_r(r,\theta)\Delta r$ |
| $B\to C$ | Increasing angle, at outer radius | $+u_\theta(r+\Delta r,\theta)(r+\Delta r)\Delta\theta$ |
| $C\to D$ | Inward, at angle $\theta+\Delta\theta$ | $-u_r(r,\theta+\Delta\theta)\Delta r$ |
| $D\to A$ | Decreasing angle, at inner radius | $-u_\theta(r,\theta)r\Delta\theta$ |

These are small-patch approximations, not exact finite-side integrals. For smooth velocity, pairing opposite sides and shrinking the patch gives the derivatives below; within-side variations do not alter the limit.

Group the two arc contributions and the two radial contributions:

$$\begin{aligned}
\Gamma\approx{}&\left[(r+\Delta r)u_\theta(r+\Delta r,\theta)-r u_\theta(r,\theta)\right]\Delta\theta\\
&-\left[u_r(r,\theta+\Delta\theta)-u_r(r,\theta)\right]\Delta r.
\end{aligned}$$

Divide by the leading patch area $r\Delta r\Delta\theta$:

$$\frac{\Gamma}{\text{area}}\approx
\frac{1}{r}\left[
\frac{(r+\Delta r)u_\theta(r+\Delta r,\theta)-r u_\theta(r,\theta)}{\Delta r}
-\frac{u_r(r,\theta+\Delta\theta)-u_r(r,\theta)}{\Delta\theta}
\right].$$

Each fraction inside the brackets is a difference quotient: change in a quantity divided by change in its coordinate. As the increments approach zero, the first becomes the radial derivative of $r u_\theta$ and the second becomes the angular derivative of $u_r$. Therefore

$$\boxed{\omega_z=\frac{1}{r}\left[\frac{\partial(r u_\theta)}{\partial r}-\frac{\partial u_r}{\partial\theta}\right].}$$

Every part now has a geometric origin:

- **$r u_\theta$ inside the derivative:** tangential velocity is multiplied by arc length, and arc length grows with radius. We compare the products on the outer and inner arcs.
- **The minus sign:** we walk outward along the lower-angle side and inward along the higher-angle side, giving lower-angle radial velocity minus higher-angle radial velocity.
- **The factor $1/r$ outside:** circulation is divided by patch area, which contains $r\Delta r\Delta\theta$.

### For this source, the arc contributions vanish

There is no tangential velocity: $u_\theta=0$. Only the two radial sides contribute. Thus

$$\omega_z=-\frac{1}{r}\frac{\partial u_r}{\partial\theta}.$$

Setting this to zero means the radial speeds on neighbouring rays cannot differ at the same radius. Consequently $u_r$ depends only on $r$.

## 5b. Apply the derived formula to source flow

**Vorticity** measures the local spinning motion of a small fluid element. In two-dimensional flow, we only need its component perpendicular to the flow plane. Call it $\omega_z$:

- $\omega$ is the Greek letter *omega*, used here for vorticity.
- The subscript $z$ labels the direction perpendicular to the $(x,y)$ plane.
- **Irrotational** means $\omega_z=0$.

The expression for this component in polar coordinates is

$$\omega_z=\frac{1}{r}\left(\frac{\partial(r u_\theta)}{\partial r}
-\frac{\partial u_r}{\partial\theta}\right).$$

This is the formula derived from the four boundary contributions above. We can now apply it to the source.

**First substitute the no-swirl condition** $u_\theta=0$. Then $r u_\theta=0$ and its derivative is also zero:

$$\omega_z=-\frac{1}{r}\frac{\partial u_r}{\partial\theta}.$$

**Now impose irrotationality**, $\omega_z=0$:

$$0=-\frac{1}{r}\frac{\partial u_r}{\partial\theta}.$$

Multiplying both sides by $-r$ is allowed because we are away from the origin:

$$\frac{\partial u_r}{\partial\theta}=0.$$

Therefore the radial velocity has no angular dependence:

$$u_r(r,\theta)=u_r(r).$$

**In words:** at a fixed distance from the source, the radial velocity is the same all the way around the circle. We have not yet found how it changes with distance.

The no-swirl assumption matters: zero vorticity alone does not force $u_\theta=0$. We set that component to zero because we are modelling a pure source with straight radial streamlines.

## 6a. Where does the polar continuity formula come from?

We can use the same patch, but now count fluid **crossing its boundary**, rather than moving along it. Outward flow is positive; inward flow is negative. We measure volume flow per unit depth, so velocity times edge length gives the appropriate flow rate.

For a small patch, the outer and inner arcs together contribute

$$\left[(r+\Delta r)u_r(r+\Delta r,\theta)-r u_r(r,\theta)\right]\Delta\theta.$$

Here it is the **radial** velocity that crosses the circular arcs. The outer arc's outward direction is positive radial; the inner arc's outward direction points toward the origin.

The two radial sides together contribute

$$\left[u_\theta(r,\theta+\Delta\theta)-u_\theta(r,\theta)\right]\Delta r.$$

Here the **tangential** velocity crosses the sides. Outward from the patch means increasing angle on its higher-angle side and decreasing angle on its lower-angle side.

For incompressible flow with no source inside this patch, total outward flow is zero. Add these contributions, divide by patch area $r\Delta r\Delta\theta$, and let the increments approach zero:

$$\frac{1}{r}\frac{\partial(r u_r)}{\partial r}
+\frac{1}{r}\frac{\partial u_\theta}{\partial\theta}=0.$$

This derives the continuity formula used next. Circulation measures velocity **along** the boundary; continuity measures velocity **across** it. That difference explains why the components appear in different places in the two formulas.

## 6b. Apply continuity to find the dependence on distance

**Incompressible** means that the fluid density stays constant. **Continuity** expresses conservation of fluid: away from the source, a small region cannot produce or absorb fluid. The total flow leaving it must balance the total flow entering it.

For incompressible two-dimensional flow, continuity in polar coordinates is

$$\frac{1}{r}\frac{\partial(r u_r)}{\partial r}
+\frac{1}{r}\frac{\partial u_\theta}{\partial\theta}=0.$$

The first term accounts for radial flow; the second accounts for tangential flow. The factor $r$ inside the radial derivative reflects the fact that a circular arc gets longer as its radius increases.

Since $u_\theta=0$, the second term is zero. Since $u_r$ depends only on $r$, we can use an ordinary derivative:

$$\frac{1}{r}\frac{d(r u_r)}{dr}=0.$$

Multiply by $r$:

$$\frac{d(r u_r)}{dr}=0.$$

A quantity whose derivative is zero does not change. Therefore **the product $r u_r$ is constant**:

$$r u_r=C.$$

Here $C$ is an unknown constant, independent of radius and angle. Dividing by $r$ gives

$$u_r(r)=\frac{C}{r}.$$

**In words:** double the distance from the source and the speed halves. Continuity determines this shape, but we still need to specify how much fluid the source supplies to determine $C$.

## 7. Source strength: find the constant C

The symbol $\sigma$ (the Greek letter *sigma*) denotes the **source strength**. In this two-dimensional model, it is the outward volume flow rate **per unit depth** perpendicular to the plane. Its units are $\mathrm{m^2/s}$.

For example, $\sigma=5\ \mathrm{m^2/s}$ corresponds to $5\ \mathrm{m^3/s}$ through a slice of depth $1\ \mathrm{m}$.

Draw a circle of radius $R$ around the source. Capital $R$ means the chosen radius of this particular circle; lowercase $r$ was the variable distance used in the velocity formula.

The fluid crosses this circle at the same outward speed $u_r(R)$ everywhere. Thus

$$\text{volume flow rate per unit depth}
=\text{radial speed}\times\text{circumference},$$

or

$$\sigma=u_r(R)\,2\pi R.$$

Substitute $u_r(R)=C/R$:

$$\sigma=\frac{C}{R}\,2\pi R=2\pi C.$$

The radius cancels, as it must: every enclosing circle carries the same flow. Solving for the constant,

$$C=\frac{\sigma}{2\pi}.$$

Substitute this into $u_r=C/r$:

$$\boxed{u_r(r,\theta)=\frac{\sigma}{2\pi r},\qquad u_\theta(r,\theta)=0,\qquad r>0.}$$

The $(r,\theta)$ notation is retained to match the lesson, although the result does not actually depend on $\theta$.

## 8. The same flow-rate calculation written as an integral

The circumference argument above is enough to finish the derivation. To connect it to integral notation, split the circle into tiny arcs.

For an angular increment $d\theta$, the corresponding arc length is $R\,d\theta$. The flow through this tiny arc per unit depth is

$$u_r(R)\,R\,d\theta.$$

The integral symbol $\int$ means to add these contributions continuously around the whole circle, from angle $0$ to $2\pi$:

$$\sigma=\int_0^{2\pi}u_r(R)\,R\,d\theta.$$

Because $u_r(R)$ and $R$ are constant on this circle,

$$\sigma=u_r(R)R\int_0^{2\pi}d\theta
=u_r(R)R(2\pi)=2\pi R u_r(R).$$

This is exactly the speed-times-circumference calculation, expressed as an integral.

## 9. Check the answer with numbers and units

Take $\sigma=5\ \mathrm{m^2/s}$:

| Distance $r$ | Radial velocity $u_r=5/(2\pi r)$ | Outward flow per unit depth $2\pi r u_r$ |
|---|---|---|
| $1\ \mathrm{m}$ | $0.7958\ \mathrm{m/s}$ | $5\ \mathrm{m^2/s}$ |
| $2\ \mathrm{m}$ | $0.3979\ \mathrm{m/s}$ | $5\ \mathrm{m^2/s}$ |
| $4\ \mathrm{m}$ | $0.1989\ \mathrm{m/s}$ | $5\ \mathrm{m^2/s}$ |

The speeds are rounded; the flow rates follow from the exact formula. A bigger circle has a larger circumference, so the same flow spreads over more space and moves more slowly.

The units also work:

$$[u_r]=\frac{\mathrm{m^2/s}}{\mathrm{m}}=\mathrm{m/s}.$$

A positive $\sigma$ gives outward velocity: a **source**. A negative $\sigma$ gives inward velocity: a **sink**.

If you want a **mass** flow rate, multiply the volume flow rate by density. With density $\rho$ (the Greek letter *rho*, in $\mathrm{kg/m^3}$), the mass flow rate per unit depth is $\rho\sigma$, in $\mathrm{kg/(m\,s)}$.

## 10. What happens at the source itself?

The formula contains $1/r$, so it is undefined at $r=0$ and its magnitude grows without bound as we approach the origin for nonzero source strength. This is an idealized **point source**, not a finite-size injection nozzle.

There is no contradiction with continuity: we applied the no-production-of-fluid equation only **away from the source**. The origin is precisely where the model supplies fluid.

Imagine a ring-shaped region between two circles around the source. Fluid enters this region across the inner circle and leaves across the outer circle at the same rate. Nothing is created inside the ring.

The derivation has three distinct steps:

1. Straight radial flow gives $u_\theta=0$; irrotationality then gives no angular dependence of $u_r$.
2. Continuity makes $r u_r$ constant, so $u_r=C/r$.
3. The prescribed source strength gives $2\pi C=\sigma$, so $u_r=\sigma/(2\pi r)$.
