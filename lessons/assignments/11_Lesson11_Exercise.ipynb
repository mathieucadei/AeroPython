&= \frac{1}{2\pi} \int_0^{l_i} \frac{\left(x_{c,i} + d \cos \beta_i - x_i + s \sin \beta_i\right) \cos \beta_i + \left(y_{c,i} + d \sin \beta_i - y_i - s \cos \beta_i\right)\sin \beta_i}{\left(x_{c,i} + d \cos \beta_i - x_i + s \sin \beta_i\right)^2 + \left(y_{c,i} + d \sin \beta_i - y_i - s \cos \beta_i\right)^2} {\rm d}s \\
&= \frac{1}{2\pi} \int_0^{l_i} \frac{\left(x_i - \frac{l_i}{2} \sin \beta_i + d \cos \beta_i - x_i + s \sin \beta_i\right) \cos \beta_i + \left(y_i + \frac{l_i}{2} \cos \beta_i + d \sin \beta_i - y_i - s \cos \beta_i\right)\sin \beta_i}{\left(x_i - \frac{l_i}{2} \sin \beta_i + d \cos \beta_i - x_i + s \sin \beta_i\right)^2 + \left(y_j + \frac{l_i}{2} \cos \beta_i + d \sin \beta_i - y_i - s \cos \beta_i\right)^2} {\rm d}s \\
&= \frac{1}{2\pi} \int_0^{l_i} \frac{\left(d \cos \beta_i + (s - \frac{l_i}{2}) \sin \beta_i \right) \cos \beta_i + \left(d \sin \beta_i - (s - \frac{l_i}{2}) \cos \beta_i \right)\sin \beta_i}{\left(d \cos \beta_i + ((s - \frac{l_i}{2}) \sin \beta_i\right))^2 + \left(d \sin \beta_i - ((s - \frac{l_i}{2}) \cos \beta_i\right))^2} {\rm d}s \\
&= \frac{1}{2\pi} \int_0^{l_i} \frac{d}{d^2 + (s - \frac{l_i}{2})^2} {\rm d}s \\
&= \frac{1}{2\pi} \left[\tan^{-1} \left(\frac{s - \frac{l_i}{2}}{d}\right)\right]_0^{l_i} \\
&= \frac{1}{2\pi} \tan^{-1} \left(\frac{l_i}{2d}\right) - \tan^{-1} \left(- \frac{l_i}{2d}\right)\\
&= \frac{1}{\pi}\tan^{-1} \left(\frac{l_i}{2d}\right) \\
&= \frac{1}{\pi} \frac{\pi}{2} = \frac{1}{2}
\end{split}
