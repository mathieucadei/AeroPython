## Part 1: Williams' test case

Now, for the 2-element airfoil, you need to derive the required mathematical expressions, following the same process as in the exercise, starting from the following expression for the potential:

$$
\begin{split}
\phi(x, y) &= U_{\infty}x\cos\alpha + U_{\infty}y\sin\alpha \\
& + \int_{main} \frac{1}{2\pi} \sigma(s) \ln \sqrt{(x-\xi(s))^2+(y-\eta(s))^2} ds  \\
& + \int_{flap} \frac{1}{2\pi} \sigma(s) \ln \sqrt{(x-\xi(s))^2+(y-\eta(s))^2} ds \\
& - \int_{main} \frac{1}{2\pi} \gamma(s) \tan^{-1} \frac{y-\eta(s)}{x-\xi(s)} ds  \\
& - \int_{flap} \frac{1}{2\pi} \gamma(s) \tan^{-1} \frac{y-\eta(s)}{x-\xi(s)} ds
\end{split}
$$


Assume the following:

1. $\sigma(s)$ is constant on each panel, 
2. $\gamma(s)$ is constant on each airfoil, and
3. the 1<sup>st</sup> to $N_a$<sup>th</sup> panels are on the main airfoil and the $(N_a+1)$<sup>th</sup> to $N$<sup>th</sup> panels are on the flap (where $N=N_a+N_b$, and $N_b$ is the number of panels on the flap).

There will be $N+2$ unknowns, that is, $\sigma_1\cdots\sigma_N$, $\gamma_a$, and $\gamma_b$. The values of $\gamma_a$, and $\gamma_b$ represent the vortex strengths on the main and flap airfoils, respectively.

$$
\begin{split}
U_n(x, y) &= \frac{\partial \phi}{\partial \vec{n}} \\
&=
U_\infty (\cos \alpha n_x + \sin \alpha n_y) \\
&+ \sum_{j=1}^{N_a} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\partial}{\partial n} \ln \left(\sqrt{(x - x_j + s \sin \beta_j)^2+(y - y_j - s \cos \beta_j)^2} \right) {\rm d}s \\
&+ \sum_{j=N_a+1}^{N} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\partial}{\partial n} \ln \left(\sqrt{(x - x_j + s \sin \beta_j)^2+(y - y_j - s \cos \beta_j)^2} \right) {\rm d}s \\
&- \frac{\gamma_a}{2\pi} \sum_{j=1}^{N_a} \int_0^{l_j} \frac{\partial}{\partial n} \tan^{-1} \left(\frac{y- y_j - s \cos \beta_j}{x - x_j + s \sin \beta_j}\right) {\rm d}s \\
&- \frac{\gamma_b}{2\pi} \sum_{j=N_a+1}^{N} \int_0^{l_j} \frac{\partial}{\partial n} \tan^{-1} \left(\frac{y- y_j - s \cos \beta_j}{x - x_j + s \sin \beta_j}\right) {\rm d}s \\
&=
U_\infty (\cos \alpha n_x + \sin \alpha n_y) \\
&+ \sum_{j=1}^{N_a} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial x}{\partial n} + \left(y - y_j - s \cos \beta_j\right)\frac{\partial y}{\partial n}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&+ \sum_{j=N_a+1}^{N} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial x}{\partial n} + \left(y - y_j - s \cos \beta_j\right)\frac{\partial y}{\partial n}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&- \frac{\gamma_a}{2\pi} \sum_{j=1}^{N_a} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial y}{\partial n} - \left(y - y_j - s \cos \beta_j\right)\frac{\partial x}{\partial n}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&- \frac{\gamma_b}{2\pi} \sum_{j=N_a+1}^{N} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial y}{\partial n} - \left(y - y_j - s \cos \beta_j\right)\frac{\partial x}{\partial n}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s
\end{split}
$$

$$
\begin{split}
U_t(x, y) &= \frac{\partial \phi}{\partial \vec{t}} \\
&=
U_\infty (\cos \alpha t_x + \sin \alpha t_y) \\
&+ \sum_{j=1}^{N_a} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\partial}{\partial t} \ln \left(\sqrt{(x - x_j + s \sin \beta_j)^2+(y - y_j - s \cos \beta_j)^2} \right) {\rm d}s \\
&+ \sum_{j=N_a+1}^{N} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\partial}{\partial t} \ln \left(\sqrt{(x - x_j + s \sin \beta_j)^2+(y - y_j - s \cos \beta_j)^2} \right) {\rm d}s \\
&- \frac{\gamma_a}{2\pi} \sum_{j=1}^{N_a} \int_0^{l_j} \frac{\partial}{\partial t} \tan^{-1} \left(\frac{y- y_j - s \cos \beta_j}{x - x_j + s \sin \beta_j}\right) {\rm d}s \\
&- \frac{\gamma_b}{2\pi} \sum_{j=N_a+1}^{N}\int_0^{l_j} \frac{\partial}{\partial t} \tan^{-1} \left(\frac{y- y_j - s \cos \beta_j}{x - x_j + s \sin \beta_j}\right) {\rm d}s \\
&=
U_\infty (\cos \alpha t_x + \sin \alpha t_y) \\
&+ \sum_{j=1}^{N_a} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial x}{\partial t} + \left(y - y_j - s \cos \beta_j\right)\frac{\partial y}{\partial t}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&+ \sum_{j=N_a+1}^{N} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial x}{\partial t} + \left(y - y_j - s \cos \beta_j\right)\frac{\partial y}{\partial t}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&- \frac{\gamma_a}{2\pi} \sum_{j=1}^{N_a} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial y}{\partial t} - \left(y - y_j - s \cos \beta_j\right)\frac{\partial x}{\partial t}}{\left(x - x_j + s \sin \beta_j \right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&- \frac{\gamma_b}{2\pi} \sum_{j=N_a+1}^{N} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial y}{\partial t} - \left(y - y_j - s \cos \beta_j\right)\frac{\partial x}{\partial t}}{\left(x - x_j + s \sin \beta_j \right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s
\end{split}
$$

\begin{split}
b^n_i &= U_\infty \cos \left(\alpha-\beta_i\right)
\end{split}

$$
\begin{split}
A^n_{ij} &= -B^t_{ij}\\
&= \frac{1}{2\pi} \int_0^{l_j} \frac{\left(x_{c,i} - x_j + s \sin \beta_j\right)\cos \beta_i + \left(y_{c,i} - y_j - s \cos \beta_j\right)\sin \beta_i}{\left(x_{c,i} - x_j + s \sin \beta_j\right)^2 + \left(y_{c,i} - y_j - s \cos \beta_j\right)^2} {\rm d}s

A^t_{ij} &= B^n_{ij} \\
&= - \frac{1}{2\pi} \int_0^{l_j} \frac{\left(x_{c,i} - x_j + s \sin \beta_j\right) \sin \beta_i - \left(y_{c,i} - y_j - s \cos \beta_j\right) \cos \beta_i}{\left(x_{c,i} - x_j + s \sin \beta_j\right)^2 + \left(y_{c,i} - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
\end{split}
$$

$$
\begin{split}
b^t_i &= U_\infty \sin \left(\alpha-\beta_i\right)
\end{split}
$$

You should be able to obtain the following matrix forms of normal and tangential velocity on the $i$<sup>th</sup> panel:

$$
U^n_i = b^n_i + 
\left[\begin{smallmatrix}A^n_{i1}\cdots A^n_{iN}\end{smallmatrix}, \sum_{j=1}^{N_a}B^n_{ij}, \sum_{j=N_a+1}^{N}B^n_{ij}\right]
\left[\begin{smallmatrix}\sigma_1 \\ \vdots \\ \sigma_N \\ \gamma_a \\ \gamma_b\end{smallmatrix}\right]
$$

$$
U^t_i = b^t_i + 
\left[\begin{smallmatrix}A^t_{i1}\cdots A^t_{iN}\end{smallmatrix}, \sum_{j=1}^{N_a}B^t_{ij}, \sum_{j=N_a+1}^{N}B^t_{ij}\right]
\left[\begin{smallmatrix}\sigma_1 \\ \vdots \\ \sigma_N \\ \gamma_a \\ \gamma_b\end{smallmatrix}\right]
$$

Using the non-penetration conditions on the $1$<sup>st</sup> to $N$<sup>th</sup> panels, you will have $N$ linear equations. To solve for the $N+2$ unknowns, the extra two linear equations you need are the Kutta conditions on the two airfoils:

$$
\begin{split}
U^t_1 &= - U^t_{N_a} \\
\left[\begin{matrix} \left(A^t_{1,1} +  A^t_{N_a,1}\right) & \left(A^t_{1,2} + A^t_{N_a,2}\right) & \dots & \left(A^t_{1,N} + A^t_{N_a,N}\right) & \left(\sum_{j=1}^{N_a} \left(B^t_{1,j} + B^t_{N_a,j}\right)\right) & \left(\sum_{j=N_a+1}^{N} \left(B^t_{1,j} + B^t_{N_a,j}\right)\right) \end{matrix}\right]\left[\begin{matrix} \sigma_1 \\ \sigma_2 \\ \vdots \\ \sigma_N \\ \gamma_a \\ \gamma_b \end{matrix}\right] &= - \left(b^t_1 + b^t_{N_a}\right)
\end{split}
$$

$$
\begin{split}
U^t_{N_a+1} &= - U^t_{N} \\
\left[\begin{matrix} \left(A^t_{N_a+1,1} +  A^t_{N,1}\right) & \left(A^t_{N_a+1,2} + A^t_{N,2}\right) & \dots & \left(A^t_{N_a+1,N} + A^t_{N,N}\right) & \left(\sum_{j=1}^{N_a} \left(B^t_{N_a+1,j} + B^t_{N,j}\right)\right) & \left(\sum_{j=N_a+1}^{N} \left(B^t_{N_a+1,j} + B^t_{N,j}\right)\right)\end{matrix}\right]\left[\begin{matrix} \sigma_1 \\ \sigma_2 \\ \vdots \\ \sigma_N \\ \gamma_a \\ \gamma_b \end{matrix}\right] &= - \left(b^t_{N_a+1} + b^t_N\right)
\end{split}
$$

The panel endpoints of each airfoil are ordered counter-clockwise, starting and ending at its trailing edge (upper surface first). The tangent of the first panel therefore points upstream and that of the last panel points downstream. Since the Kutta condition requires equal flow speed leaving the trailing edge on both sides, the tangential components have opposite signs: $U^t_1 = -U^t_{N_a}$ and $U^t_{N_a+1} = -U^t_N$. The hint's $U^t_1 = U^t_{N_a}$ should be read as equality of magnitudes, $\left|U^t_1\right| = \left|U^t_{N_a}\right|$.

$$
\begin{split}
&\left[\begin{matrix} 
A^n_{1,1} & A^n_{1,2} & \dots & A^n_{1,N} & \left(\sum_{j=1}^{N_a} B^n_{1,j}\right) & \left(\sum_{j=N_a+1}^{N} B^n_{1,j}\right) \\
A^n_{2,1} & A^n_{2,2} & \dots & A^n_{2,N} & \left(\sum_{j=1}^{N_a} B^n_{2,j}\right) & \left(\sum_{j=N_a+1}^{N} B^n_{2,j}\right) \\
\vdots & \vdots & \ddots & \vdots & \vdots & \vdots \\
A^n_{N,1} & A^n_{N,2} & \dots & A^n_{N,N} & \left(\sum_{j=1}^{N_a} B^n_{N,j}\right) & \left(\sum_{j=N_a+1}^{N} B^n_{N,j}\right) \\
\left(A^t_{1,1} +  A^t_{N_a,1}\right) & \left(A^t_{1,2} + A^t_{N_a,2}\right) & \dots & \left(A^t_{1,N} + A^t_{N_a,N}\right) & \left(\sum_{j=1}^{N_a} \left(B^t_{1,j} + B^t_{N_a,j}\right)\right) & \left(\sum_{j=N_a+1}^N \left(B^t_{1,j} + B^t_{N_a,j}\right)\right) \\
\left(A^t_{N_a+1,1} +  A^t_{N,1}\right) & \left(A^t_{N_a+1,2} + A^t_{N,2}\right) & \dots & \left(A^t_{N_a+1,N} + A^t_{N,N}\right) & \left(\sum_{j=1}^{N_a} \left(B^t_{N_a+1,j} + B^t_{N,j}\right)\right) & \left(\sum_{j=N_a+1}^N \left(B^t_{N_a+1,j} + B^t_{N,j}\right)\right) \\
\end{matrix}\right]\left[\begin{matrix} \sigma_1 \\ \sigma_2 \\ \vdots \\ \sigma_N \\ \gamma_a \\ \gamma_b \end{matrix}\right] = \left[\begin{matrix} - b_1^n \\ - b_2^n \\ \vdots \\ - b_N^n \\ - \left(b^t_1 + b^t_{N_a}\right) \\ - \left(b^t_{N_a+1} + b^t_{N}\right) \end{matrix}\right]\\
\end{split}
$$

```python
class Panel:
    """
    Contains information related to a panel.
    """
    def __init__(self, xa, ya, xb, yb):
        """
        Initializes the panel.
        
        Sets the end-points and calculates the center-point, length,
        and angle (with the x-axis) of the panel.
        Defines if the panel is located on the upper or lower surface of the geometry.
        Initializes the source-strength, tangential velocity, and pressure coefficient
        of the panel to zero.
        
        Parameters
        ----------
        xa: float
            x-coordinate of the first end-point.
        ya: float
            y-coordinate of the first end-point.
        xb: float
            x-coordinate of the second end-point.
        yb: float
            y-coordinate of the second end-point.
        """
        self.xa, self.ya = xa, ya  # panel starting-point
        self.xb, self.yb = xb, yb  # panel ending-point
        
        self.xc, self.yc = (xa + xb) / 2, (ya + yb) / 2  # panel center
        self.length = np.sqrt((xb - xa)**2 + (yb - ya)**2)  # panel length
        
        # orientation of panel (angle between x-axis and panel's normal)
        if xb - xa <= 0.0:
            self.beta = np.arccos((yb - ya) / self.length)
        elif xb - xa > 0.0:
            self.beta = np.pi + np.arccos(-(yb - ya) / self.length)
        
        # panel location
        if self.beta <= np.pi:
            self.loc = 'upper'  # upper surface
        else:
            self.loc = 'lower'  # lower surface
        
        self.sigma = 0.0  # source strength
        self.vt = 0.0  # tangential velocity
        self.cp = 0.0  # pressure coefficient
```

```python
class Freestream:
    """
    Freestream conditions.
    """
    def __init__(self, U_inf=1.0, alpha=0.0):
        """
        Sets the freestream speed and angle (in degrees).
        
        Parameters
        ----------
        U_inf: float, optional
            Freestream speed;
            default: 1.0.
        alpha: float, optional
            Angle of attack in degrees;
            default 0.0.
        """
        self.U_inf = U_inf
        self.alpha = np.radians(alpha)  # degrees to radians
```

```python
def integral(x, y, panel, dxdk, dydk):
    """
    Evaluates the contribution from a panel at a given point.
    
    Parameters
    ----------
    x: float
        x-coordinate of the target point.
    y: float
        y-coordinate of the target point.
    panel: Panel object
        Panel whose contribution is evaluated.
    dxdk: float
        Value of the derivative of x in a certain direction.
    dydk: float
        Value of the derivative of y in a certain direction.
    
    Returns
    -------
    Contribution from the panel at a given point (x, y).
    """
    def integrand(s):
        return (((x - (panel.xa - np.sin(panel.beta) * s)) * dxdk +
                 (y - (panel.ya + np.cos(panel.beta) * s)) * dydk) /
                ((x - (panel.xa - np.sin(panel.beta) * s))**2 +
                 (y - (panel.ya + np.cos(panel.beta) * s))**2) )
    return integrate.quad(integrand, 0.0, panel.length)[0]
```

```python
def source_contribution_normal(panels):
    """
    Builds the source contribution matrix for the normal velocity.
    
    Parameters
    ----------
    panels: 1D array of Panel objects
        List of panels.
    
    Returns
    -------
    A: 2D Numpy array of floats
        Source contribution matrix.
    """
    A = np.empty((panels.size, panels.size), dtype=float)
    # source contribution on a panel from itself
    np.fill_diagonal(A, 0.5)
    # source contribution on a panel from others
    for i, panel_i in enumerate(panels):
        for j, panel_j in enumerate(panels):
            if i != j:
                A[i, j] = 0.5 / np.pi * integral(panel_i.xc, panel_i.yc, 
                                                    panel_j,
                                                    np.cos(panel_i.beta),
                                                    np.sin(panel_i.beta))
    return A
```

```python
def vortex_contribution_normal(panels):
    """
    Builds the vortex contribution matrix for the normal velocity.
    
    Parameters
    ----------
    panels: 1D array of Panel objects
        List of panels.
    
    Returns
    -------
    A: 2D Numpy array of floats
        Vortex contribution matrix.
    """
    A = np.empty((panels.size, panels.size), dtype=float)
    # vortex contribution on a panel from itself
    np.fill_diagonal(A, 0.0)
    # vortex contribution on a panel from others
    for i, panel_i in enumerate(panels):
        for j, panel_j in enumerate(panels):
            if i != j:
                A[i, j] = -0.5 / np.pi * integral(panel_i.xc, panel_i.yc, 
                                                     panel_j,
                                                     np.sin(panel_i.beta),
                                                     -np.cos(panel_i.beta))
    return A
```

```python
multielement_panels = np.concatenate((mainfoil_panels, flapfoil_panels))

A_source = source_contribution_normal(multielement_panels)
B_vortex = vortex_contribution_normal(multielement_panels)
```

```python
def kutta_condition(A_source, B_vortex, N):
    """
    Builds the Kutta condition array.
    
    Parameters
    ----------
    A_source: 2D Numpy array of floats
        Source contribution matrix.
    B_vortex: 2D Numpy array of floats
        Vortex contribution matrix.
    N: int
        Number of panels on the main airfoil.
    
    Returns
    -------
    b: 2D Numpy array of floats
        The left-hand side of the Kutta-condition equation.
    """

    b = np.empty((2, A_source.shape[0] + 2), dtype=float)
    # matrix of source contribution on tangential velocity
    # is the same than
    # matrix of vortex contribution on normal velocity
    b[0, :-2] = B_vortex[0, :] + B_vortex[N-1, :]
    b[1, :-2] = B_vortex[N, :] + B_vortex[-1, :]   
    # matrix of vortex contribution on tangential velocity
    # is the opposite of
    # matrix of source contribution on normal velocity
    b[0, -2] = - np.sum(A_source[0, :N] + A_source[N-1, :N])
    b[0, -1] = - np.sum(A_source[0, N:] + A_source[N-1, N:])
    b[1, -2] = - np.sum(A_source[N, :N] + A_source[-1, :N])
    b[1, -1] = - np.sum(A_source[N, N:] + A_source[-1, N:])
    return b
```

```python
def build_singularity_matrix(A_source, B_vortex, N):
    """
    Builds the left-hand side matrix of the system
    arising from source and vortex contributions.
    
    Parameters
    ----------
    A_source: 2D Numpy array of floats
        Source contribution matrix.
    B_vortex: 2D Numpy array of floats
        Vortex contribution matrix.
    N: int
        Number of panels on the main airfoil.

    Returns
    -------
    A:  2D Numpy array of floats
        Matrix of the linear system.
    """

    A = np.empty((A_source.shape[0] + 2, A_source.shape[1] + 2), dtype=float)
    # source contribution matrix
    A[:-2, :-2] = A_source
    # vortex contribution array
    A[:-2, -2] = np.sum(B_vortex[:, :N], axis=1)
    A[:-2, -1] = np.sum(B_vortex[:, N:], axis=1)
    # Kutta condition array
    A[-2:, :] = kutta_condition(A_source, B_vortex, N)
    return A
```

```python
def build_freestream_rhs(panels, freestream, N):
    """
    Builds the right-hand side of the system 
    arising from the freestream contribution.
    
    Parameters
    ----------
    panels: 1D array of Panel objects
        List of panels.
    freestream: Freestream object
        Freestream conditions.
    N: int
        Number of panels on the main airfoil.
    
    Returns
    -------
    b: 1D Numpy array of floats
        Freestream contribution on each panel and on the Kutta condition.
    """
    b = np.empty(panels.size + 2, dtype=float)
    # freestream contribution on each panel
    for i, panel in enumerate(panels):
        b[i] = -freestream.U_inf * np.cos(freestream.alpha - panel.beta)
    # freestream contribution on the Kutta condition
    b[-2] = -freestream.U_inf * (np.sin(freestream.alpha - panels[0].beta) +
                                 np.sin(freestream.alpha - panels[N - 1].beta) )
    b[-1] = -freestream.U_inf * (np.sin(freestream.alpha - panels[N].beta) +
                                 np.sin(freestream.alpha - panels[-1].beta) )
    return b
```

```python
A = build_singularity_matrix(A_source, B_vortex, N)
b = build_freestream_rhs(multielement_panels, freestream, N)
```

```python
# solve for singularity strengths
strengths = np.linalg.solve(A, b)

# store source strength on each panel
for i , panel in enumerate(multielement_panels):
    panel.sigma = strengths[i]
    
# store circulation density
gamma_a = strengths[-2]
gamma_b = strengths[-1]
```

```python
def compute_tangential_velocity(panels, freestream, gamma_a, gamma_b, A_source, B_vortex, N):
    """
    Computes the tangential surface velocity.
    
    Parameters
    ----------
    panels: 1D array of Panel objects
        List of panels.
    freestream: Freestream object
        Freestream conditions.
    gamma_a: float
        Circulation density for the first vortex.
    gamma_b: float
        Circulation density for the second vortex.
    A_source: 2D Numpy array of floats
        Source contribution matrix for the normal velocity.
    B_vortex: 2D Numpy array of floats
        Vortex contribution matrix for the normal velocity.
    N: int
        Number of panels on the main airfoil.
    """
    A = np.empty((panels.size, panels.size + 2), dtype=float)
    # matrix of source contribution on tangential velocity
    # is the same than
    # matrix of vortex contribution on normal velocity
    A[:, :-2] = B_vortex
    # matrix of vortex contribution on tangential velocity
    # is the opposite of
    # matrix of source contribution on normal velocity
    A[:, -2] = -np.sum(A_source[:, :N], axis=1)
    A[:, -1] = -np.sum(A_source[:, N:], axis=1)
    # freestream contribution
    b = freestream.U_inf * np.sin([freestream.alpha - panel.beta 
                                      for panel in panels])
    
    strengths = np.append([panel.sigma for panel in panels], [gamma_a, gamma_b])
    
    tangential_velocities = np.dot(A, strengths) + b
    
    for i, panel in enumerate(panels):
        panel.vt = tangential_velocities[i]
```

```python
# tangential velocity at each panel center.
compute_tangential_velocity(multielement_panels, freestream, gamma_a, gamma_b, A_source, B_vortex, N)
```

```python
def compute_pressure_coefficient(panels, freestream):
    """
    Computes the surface pressure coefficients.
    
    Parameters
    ----------
    panels: 1D array of Panel objects
        List of panels.
    freestream: Freestream object
        Freestream conditions.
    """
    for panel in panels:
        panel.cp = 1.0 - (panel.vt / freestream.U_inf)**2
```

```python
def rotate_flap(flap_deflection, hinge_x, hinge_y, flap_x, flap_y):
    """
    Rotates the flap coordinates about the hinge point.
    
    A positive deflection rotates the flap clockwise (trailing edge down).
    
    Parameters
    ----------
    flap_deflection: float
        Flap deflection angle in degrees.
    hinge_x: float
        x-coordinate of the hinge.
    hinge_y: float
        y-coordinate of the hinge.
    flap_x: 1D array of floats
        x-coordinates of the flap points.
    flap_y: 1D array of floats
        y-coordinates of the flap points.
    
    Returns
    -------
    x_flap_rotated: 1D array of floats
        x-coordinates of the rotated flap points.
    y_flap_rotated: 1D array of floats
        y-coordinates of the rotated flap points.
    """
    
    theta = np.radians(flap_deflection)

    x = flap_x - hinge_x
    y = flap_y - hinge_y

    x_prime = x*np.cos(theta) + y * np.sin(theta)
    y_prime = -x * np.sin(theta) + y * np.cos(theta)

    x_flap_rotated = x_prime + hinge_x
    y_flap_rotated = y_prime + hinge_y

    return x_flap_rotated, y_flap_rotated
```

```python
def kutta_condition_single_element(A_source, B_vortex):
    """
    Builds the Kutta condition array.
    
    Parameters
    ----------
    A_source: 2D Numpy array of floats
        Source contribution matrix for the normal velocity.
    B_vortex: 2D Numpy array of floats
        Vortex contribution matrix for the normal velocity.
    
    Returns
    -------
    b: 1D Numpy array of floats
        The left-hand side of the Kutta-condition equation.
    """
    b = np.empty(A_source.shape[0] + 1, dtype=float)
    # matrix of source contribution on tangential velocity
    # is the same than
    # matrix of vortex contribution on normal velocity
    b[:-1] = B_vortex[0, :] + B_vortex[-1, :]
    # matrix of vortex contribution on tangential velocity
    # is the opposite of
    # matrix of source contribution on normal velocity
    b[-1] = - np.sum(A_source[0, :] + A_source[-1, :])
    return b
```

```python
def build_singularity_matrix_single_element(A_source, B_vortex):
    """
    Builds the left-hand side matrix of the system
    arising from source and vortex contributions.
    
    Parameters
    ----------
    A_source: 2D Numpy array of floats
        Source contribution matrix for the normal velocity.
    B_vortex: 2D Numpy array of floats
        Vortex contribution matrix for the normal velocity.
    
    Returns
    -------
    A:  2D Numpy array of floats
        Matrix of the linear system.
    """
    A = np.empty((A_source.shape[0] + 1, A_source.shape[1] + 1), dtype=float)
    # source contribution matrix
    A[:-1, :-1] = A_source
    # vortex contribution array
    A[:-1, -1] = np.sum(B_vortex, axis=1)
    # Kutta condition array
    A[-1, :] = kutta_condition_single_element(A_source, B_vortex)
    return A
```

```python
def build_freestream_rhs_single_element(panels, freestream):
    """
    Builds the right-hand side of the system 
    arising from the freestream contribution.
    
    Parameters
    ----------
    panels: 1D array of Panel objects
        List of panels.
    freestream: Freestream object
        Freestream conditions.
    
    Returns
    -------
    b: 1D Numpy array of floats
        Freestream contribution on each panel and on the Kutta condition.
    """
    b = np.empty(panels.size + 1, dtype=float)
    # freestream contribution on each panel
    for i, panel in enumerate(panels):
        b[i] = -freestream.U_inf * np.cos(freestream.alpha - panel.beta)
    # freestream contribution on the Kutta condition
    b[-1] = -freestream.U_inf * (np.sin(freestream.alpha - panels[0].beta) +
                                 np.sin(freestream.alpha - panels[-1].beta) )
    return b
```

```python
def compute_tangential_velocity_single_element(panels, freestream, gamma, A_source, B_vortex):
    """
    Computes the tangential surface velocity.
    
    Parameters
    ----------
    panels: 1D array of Panel objects
        List of panels.
    freestream: Freestream object
        Freestream conditions.
    gamma: float
        Circulation density.
    A_source: 2D Numpy array of floats
        Source contribution matrix for the normal velocity.
    B_vortex: 2D Numpy array of floats
        Vortex contribution matrix for the normal velocity.
    """
    A = np.empty((panels.size, panels.size + 1), dtype=float)
    # matrix of source contribution on tangential velocity
    # is the same than
    # matrix of vortex contribution on normal velocity
    A[:, :-1] = B_vortex
    # matrix of vortex contribution on tangential velocity
    # is the opposite of
    # matrix of source contribution on normal velocity
    A[:, -1] = -np.sum(A_source, axis=1)
    # freestream contribution
    b = freestream.U_inf * np.sin([freestream.alpha - panel.beta 
                                      for panel in panels])
    
    strengths = np.append([panel.sigma for panel in panels], gamma)
    
    tangential_velocities = np.dot(A, strengths) + b
    
    for i, panel in enumerate(panels):
        panel.vt = tangential_velocities[i]
```

```python
def run_single_element_analysis(
        mainfoil_file, 
        freestream_u_inf, 
        freestream_alpha
    ):
    """
    Runs the vortex-source panel method on a single airfoil.
    
    Builds the panels, solves for the source strengths and the vortex strength,
    then computes the tangential velocity and pressure coefficient on each panel.
    
    Parameters
    ----------
    mainfoil_file: string
        Path to the CSV file with the airfoil coordinates.
    freestream_u_inf: float
        Freestream speed.
    freestream_alpha: float
        Angle of attack in degrees.
    
    Returns
    -------
    x_mainfoil: 1D array of floats
        x-coordinates of the airfoil.
    y_mainfoil: 1D array of floats
        y-coordinates of the airfoil.
    mainfoil_panels: 1D array of Panel objects
        Panels of the airfoil, with sigma, vt and cp computed.
    """

    mainfoil = pd.read_csv(mainfoil_file, header=None)

    x_mainfoil = mainfoil[0]
    y_mainfoil = mainfoil[1]
    N = len(x_mainfoil[:-1])
    
    # create panels
    mainfoil_panels = np.empty(N, dtype=object)
    for i in range(N):
        mainfoil_panels[i] = Panel(x_mainfoil[i], y_mainfoil[i], x_mainfoil[i + 1], y_mainfoil[i + 1])

    A_source = source_contribution_normal(mainfoil_panels)
    B_vortex = vortex_contribution_normal(mainfoil_panels)

    freestream = Freestream(U_inf=freestream_u_inf, alpha=freestream_alpha)

    A = build_singularity_matrix_single_element(A_source, B_vortex)
    b = build_freestream_rhs_single_element(mainfoil_panels, freestream)

    # solve for singularity strengths
    strengths = np.linalg.solve(A, b)

    # store source strength on each panel
    for i , panel in enumerate(mainfoil_panels):
        panel.sigma = strengths[i]
        
    # store circulation density
    gamma = strengths[-1]

    # tangential velocity at each panel center.
    compute_tangential_velocity_single_element(mainfoil_panels, freestream, gamma, A_source, B_vortex)

    # surface pressure coefficient
    compute_pressure_coefficient(mainfoil_panels, freestream)

    return x_mainfoil, y_mainfoil, mainfoil_panels
```

```python
def run_multielement_analysis(
        mainfoil_file, 
        flapfoil_file,
        hinge_x,
        hinge_y,
        flap_deflection,
        freestream_u_inf, 
        freestream_alpha
    ):
    """
    Runs the vortex-source panel method on the main airfoil with a deflected flap.
    
    Rotates the flap about the hinge, builds the panels of both airfoils,
    solves for the source strengths and the two vortex strengths, then computes
    the tangential velocity and pressure coefficient on each panel.
    
    Parameters
    ----------
    mainfoil_file: string
        Path to the CSV file with the main airfoil coordinates.
    flapfoil_file: string
        Path to the CSV file with the flap coordinates.
    hinge_x: float
        x-coordinate of the flap hinge.
    hinge_y: float
        y-coordinate of the flap hinge.
    flap_deflection: float
        Flap deflection angle in degrees.
    freestream_u_inf: float
        Freestream speed.
    freestream_alpha: float
        Angle of attack in degrees.
    
    Returns
    -------
    x_mainfoil: 1D array of floats
        x-coordinates of the main airfoil.
    y_mainfoil: 1D array of floats
        y-coordinates of the main airfoil.
    x_flap_deflected: 1D array of floats
        x-coordinates of the deflected flap.
    y_flap_deflected: 1D array of floats
        y-coordinates of the deflected flap.
    multielement_panels: 1D array of Panel objects
        Panels of the main airfoil followed by the panels of the flap,
        with sigma, vt and cp computed.
    """
    
    mainfoil = pd.read_csv(mainfoil_file, header=None)
    flapfoil = pd.read_csv(flapfoil_file, header=None)

    x_mainfoil = mainfoil[0]
    y_mainfoil = mainfoil[1]
    x_flapfoil = flapfoil[0]
    y_flapfoil = flapfoil[1]
    N = len(x_mainfoil[:-1])

    x_flap_deflected, y_flap_deflected = rotate_flap(flap_deflection, hinge_x, hinge_y, x_flapfoil, y_flapfoil)
    
    # create panels
    mainfoil_panels = np.empty(N, dtype=object)
    for i in range(N):
        mainfoil_panels[i] = Panel(x_mainfoil[i], y_mainfoil[i], x_mainfoil[i + 1], y_mainfoil[i + 1])

    # create panels
    flapfoil_panels = np.empty(N, dtype=object)
    for i in range(N):
        flapfoil_panels[i] = Panel(x_flap_deflected[i], y_flap_deflected[i], x_flap_deflected[i + 1], y_flap_deflected[i + 1])

    multielement_panels = np.concatenate((mainfoil_panels, flapfoil_panels))

    A_source = source_contribution_normal(multielement_panels)
    B_vortex = vortex_contribution_normal(multielement_panels)

    freestream = Freestream(U_inf=freestream_u_inf, alpha=freestream_alpha)

    A = build_singularity_matrix(A_source, B_vortex, N)
    b = build_freestream_rhs(multielement_panels, freestream, N)

    # solve for singularity strengths
    strengths = np.linalg.solve(A, b)

    # store source strength on each panel
    for i , panel in enumerate(multielement_panels):
        panel.sigma = strengths[i]
        
    # store circulation density
    gamma_a = strengths[-2]
    gamma_b = strengths[-1]

    # tangential velocity at each panel center.
    compute_tangential_velocity(multielement_panels, freestream, gamma_a, gamma_b, A_source, B_vortex, N)

    # surface pressure coefficient
    compute_pressure_coefficient(multielement_panels, freestream)

    return x_mainfoil, y_mainfoil, x_flap_deflected, y_flap_deflected, multielement_panels
```

```python
def lift_at_alpha(freestream_alpha, flap_deflection):
    """
    Computes the lift for a given angle of attack and flap deflection.
    
    Uses the global constants mainfoil_file, flapfoil_file, hinge_x, hinge_y
    and freestream_u_inf.
    
    Parameters
    ----------
    freestream_alpha: float
        Angle of attack in degrees.
    flap_deflection: float
        Flap deflection angle in degrees.
    
    Returns
    -------
    float
        Lift per unit span, perpendicular to the freestream,
        non-dimensionalized by dynamic pressure.
    """
    
    *_, panels = run_multielement_analysis(
        mainfoil_file,
        flapfoil_file,
        hinge_x,
        hinge_y,
        flap_deflection,
        freestream_u_inf,
        freestream_alpha,
    )

    # Pressure-force components along the fixed x and y axes
    force_x = -sum(p.cp * p.length * np.cos(p.beta) for p in panels)
    force_y = -sum(p.cp * p.length * np.sin(p.beta) for p in panels)

    # Lift acts perpendicular to the freestream
    angle = np.radians(freestream_alpha)
    return -force_x * np.sin(angle) + force_y * np.cos(angle)

flap_deflection = 20
zero_lift_alpha = optimize.bisect(lift_at_alpha, -14, 14, args=(flap_deflection))
```

```python
def lift_at_flap_deflection(flap_deflection, freestream_alpha):
    """
    Computes the lift for a given flap deflection and angle of attack.
    
    Uses the global constants mainfoil_file, flapfoil_file, hinge_x, hinge_y
    and freestream_u_inf.
    
    Parameters
    ----------
    flap_deflection: float
        Flap deflection angle in degrees.
    freestream_alpha: float
        Angle of attack in degrees.
    
    Returns
    -------
    float
        Lift per unit span, perpendicular to the freestream,
        non-dimensionalized by dynamic pressure.
    """
    
    *_, panels = run_multielement_analysis(
        mainfoil_file,
        flapfoil_file,
        hinge_x,
        hinge_y,
        flap_deflection,
        freestream_u_inf,
        freestream_alpha,
    )

    # Pressure-force components along the fixed x and y axes
    force_x = -sum(p.cp * p.length * np.cos(p.beta) for p in panels)
    force_y = -sum(p.cp * p.length * np.sin(p.beta) for p in panels)

    # Lift acts perpendicular to the freestream
    angle = np.radians(freestream_alpha)
    return -force_x * np.sin(angle) + force_y * np.cos(angle)
```
