Now, for the 2-element airfoil, you need to derive the required mathematical expressions, following the same process as in the exercise, starting from the following expression for the potential:

$$
\begin{split}
\phi(x, y) &= U_{\infty}x\cos\alpha + U_{\infty}y\sin\alpha \\
& + \int_{main} \frac{1}{2\pi} \sigma(s) \ln \sqrt{(x-\xi(s))^2+(y-\eta(s))^2} ds  \\
& + \int_{flap} \frac{1}{2\pi} \sigma(s) \ln \sqrt{(x-\xi(s))^2+(y-\eta(s))^2} ds \\
& - \int_{main} \frac{1}{2\pi} \gamma(s) \tan^{-1} \frac{y-\eta(s)}{x-\xi(s)} ds  \\
& - \int_{flap} \frac{1}{2\pi} \gamma(s) \tan^{-1} \frac{y-\eta(s)}{x-\xi(s)} ds
\end{split}
$$


Assume the following:

1. $\sigma(s)$ is constant on each panel, 
2. $\gamma(s)$ is constant on each airfoil, and
3. the 1<sup>st</sup> to $N_a$<sup>th</sup> panels are on the main airfoil and the $(N_a+1)$<sup>th</sup> to $N$<sup>th</sup> panels are on the flap (where $N=N_a+N_b$, and $N_b$ is the number of panels on the flap).

There will be $N+2$ unknowns, that is, $\sigma_1\cdots\sigma_N$, $\gamma_a$, and $\gamma_b$. The values of $\gamma_a$, and $\gamma_b$ represent the vortex strengths on the main and flap airfoils, respectively.

$$
\begin{split}
U_n(x, y) &= \frac{\partial \phi}{\partial \vec{n}} \\
&=
U_\infty (\cos \alpha n_x + \sin \alpha n_y) \\
&+ \sum_{j=1}^{N_a} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\partial}{\partial n} \ln \left(\sqrt{(x - x_j + s \sin \beta_j)^2+(y - y_j - s \cos \beta_j)^2} \right) {\rm d}s \\
&+ \sum_{j=N_a+1}^{N} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\partial}{\partial n} \ln \left(\sqrt{(x - x_j + s \sin \beta_j)^2+(y - y_j - s \cos \beta_j)^2} \right) {\rm d}s \\
&- \frac{\gamma_a}{2\pi} \sum_{j=1}^{N_a} \int_0^{l_j} \frac{\partial}{\partial n} \tan^{-1} \left(\frac{y- y_j - s \cos \beta_j}{x - x_j + s \sin \beta_j}\right) {\rm d}s \\
&- \frac{\gamma_b}{2\pi} \sum_{j=N_a+1}^{N} \int_0^{l_j} \frac{\partial}{\partial n} \tan^{-1} \left(\frac{y- y_j - s \cos \beta_j}{x - x_j + s \sin \beta_j}\right) {\rm d}s \\
&=
U_\infty (\cos \alpha n_x + \sin \alpha n_y) \\
&+ \sum_{j=1}^{N_a} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial x}{\partial n} + \left(y - y_j - s \cos \beta_j\right)\frac{\partial y}{\partial n}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&+ \sum_{j=N_a+1}^{N} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial x}{\partial n} + \left(y - y_j - s \cos \beta_j\right)\frac{\partial y}{\partial n}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&- \frac{\gamma_a}{2\pi} \sum_{j=1}^{N_a} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial y}{\partial n} - \left(y - y_j - s \cos \beta_j\right)\frac{\partial x}{\partial n}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&- \frac{\gamma_b}{2\pi} \sum_{j=N_a+1}^{N} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial y}{\partial n} - \left(y - y_j - s \cos \beta_j\right)\frac{\partial x}{\partial n}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s
\end{split}
$$

$$
\begin{split}
U_t(x, y) &= \frac{\partial \phi}{\partial \vec{t}} \\
&=
U_\infty (\cos \alpha t_x + \sin \alpha t_y) \\
&+ \sum_{j=1}^{N_a} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\partial}{\partial t} \ln \left(\sqrt{(x - x_j + s \sin \beta_j)^2+(y - y_j - s \cos \beta_j)^2} \right) {\rm d}s \\
&+ \sum_{j=N_a+1}^{N} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\partial}{\partial t} \ln \left(\sqrt{(x - x_j + s \sin \beta_j)^2+(y - y_j - s \cos \beta_j)^2} \right) {\rm d}s \\
&- \frac{\gamma_a}{2\pi} \sum_{j=1}^{N_a} \int_0^{l_j} \frac{\partial}{\partial t} \tan^{-1} \left(\frac{y- y_j - s \cos \beta_j}{x - x_j + s \sin \beta_j}\right) {\rm d}s \\
&- \frac{\gamma_b}{2\pi} \sum_{j=N_a+1}^{N}\int_0^{l_j} \frac{\partial}{\partial t} \tan^{-1} \left(\frac{y- y_j - s \cos \beta_j}{x - x_j + s \sin \beta_j}\right) {\rm d}s \\
&=
U_\infty (\cos \alpha t_x + \sin \alpha t_y) \\
&+ \sum_{j=1}^{N_a} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial x}{\partial t} + \left(y - y_j - s \cos \beta_j\right)\frac{\partial y}{\partial t}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&+ \sum_{j=N_a+1}^{N} \frac{\sigma_j}{2\pi} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial x}{\partial t} + \left(y - y_j - s \cos \beta_j\right)\frac{\partial y}{\partial t}}{\left(x - x_j + s \sin \beta_j\right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&- \frac{\gamma_a}{2\pi} \sum_{j=1}^{N_a} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial y}{\partial t} - \left(y - y_j - s \cos \beta_j\right)\frac{\partial x}{\partial t}}{\left(x - x_j + s \sin \beta_j \right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s \\
&- \frac{\gamma_b}{2\pi} \sum_{j=N_a+1}^{N} \int_0^{l_j} \frac{\left(x - x_j + s \sin \beta_j\right)\frac{\partial y}{\partial t} - \left(y - y_j - s \cos \beta_j\right)\frac{\partial x}{\partial t}}{\left(x - x_j + s \sin \beta_j \right)^2 + \left(y - y_j - s \cos \beta_j\right)^2} {\rm d}s
\end{split}
$$

$$
\begin{split}
b^n_i &= U_\infty \cos \left(\alpha-\beta_i\right)
\end{split}

$$
\begin{split}
B^t_{ij} &= -A^n_{ij}\\
&= -\frac{1}{2\pi} \int_0^{l_j} \frac{\left(x_{c,i} - x_j + s \sin \beta_j\right)\cos \beta_i + \left(y_{c,i} - y_j - s \cos \beta_j\right)\sin \beta_i}{\left(x_{c,i} - x_j + s \sin \beta_j\right)^2 + \left(y_{c,i} - y_j - s \cos \beta_j\right)^2} {\rm d}s
\end{split}
$$

B^n_{ij} &= A^t_{ij} \\

\end{split}

b^t_i &= U_\infty \sin \left(\alpha-\beta_i\right)

You should be able to obtain the following matrix forms of normal and tangential velocity on the $i$<sup>th</sup> panel:

$$
U^n_i = b^n_i + 
\left[\begin{smallmatrix}A^n_{i1}\cdots A^n_{iN}\end{smallmatrix}, \sum_{j=1}^{N_a}B^n_{ij}, \sum_{j=N_a+1}^{N}B^n_{ij}\right]
\left[\begin{smallmatrix}\sigma_1 \\ \vdots \\ \sigma_N \\ \gamma_a \\ \gamma_b\end{smallmatrix}\right]
$$

$$
U^t_i = b^t_i + 
\left[\begin{smallmatrix}A^t_{i1}\cdots A^t_{iN}\end{smallmatrix}, \sum_{j=1}^{N_a}B^t_{ij}, \sum_{j=N_a+1}^{N}B^t_{ij}\right]
\left[\begin{smallmatrix}\sigma_1 \\ \vdots \\ \sigma_N \\ \gamma_a \\ \gamma_b\end{smallmatrix}\right]
$$

Using the non-penetration conditions on the $1$<sup>st</sup> to $N$<sup>th</sup> panels, you will have $N$ linear equations. To solve for the $N+2$ unknowns, the extra two linear equations you need are the Kutta conditions on the two airfoils:

$$
\begin{split}
U^t_1 &= - U^t_{N_a} \\
\left[\begin{matrix} \left(A^t_{11} +  A^t_{N_a1}\right) & \left(A^t_{12} + A^t_{N_a2}\right) & \dots & \left(A^t_{1N} + A^t_{N_aN}\right) & \left(\sum_{j=1}^{N_a} \left(B^t_{1j} + B^t_{Nj}\right)\right) & \left(\sum_{j=N_a+1}^{N} \left(B^t_{1j} + B^t_{Nj}\right)\right) \end{matrix}\right]\left[\begin{matrix} \sigma_1 \\ \sigma_2 \\ \vdots \\ \sigma_N \\ \gamma_a \\ \gamma_b \end{matrix}\right] &= - \left(b^t_1 + b^t_N\right)
\end{split}
$$

$$
\begin{split}
U^t_{N_a+1} &= - U^t_{N} \\
\left[\begin{matrix} \left(A^t_{11} +  A^t_{N_a+1,1}\right) & \left(A^t_{12} + A^t_{N_a+1,2}\right) & \dots & \left(A^t_{1N} + A^t_{N_a+1,N}\right) & \left(\sum_{j=1}^{N_a} \left(B^t_{1j} + B^t_{Nj}\right)\right) & \left(\sum_{j=N_a+1}^{N} \left(B^t_{1j} + B^t_{Nj}\right)\right)\end{matrix}\right]\left[\begin{matrix} \sigma_1 \\ \sigma_2 \\ \vdots \\ \sigma_N \\ \gamma_a \\ \gamma_b \end{matrix}\right] &= - \left(b^t_1 + b^t_N\right)
\end{split}
$$

The panel endpoints of each airfoil are ordered counter-clockwise, starting and ending at its trailing edge (upper surface first). The tangent of the first panel therefore points upstream and that of the last panel points downstream. Since the Kutta condition requires equal flow speed leaving the trailing edge on both sides, the tangential components have opposite signs: $U^t_1 = -U^t_{N_a}$ and $U^t_{N_a+1} = -U^t_N$. The hint's $U^t_1 = U^t_{N_a}$ should be read as equality of magnitudes, $\left|U^t_1\right| = \left|U^t_{N_a}\right|$.
